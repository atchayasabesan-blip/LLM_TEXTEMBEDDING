```python
sentences = [
    "Artificial intelligence is transforming many industries around the world.",
    "Machine learning allows computers to learn patterns from data.",
    "Researchers are developing new technologies for renewable energy.",
    "Solar power is becoming an important source of clean electricity.",
    "The government announced new measures to improve public education.",
    "Online learning provides students with access to educational resources.",
    "Scientists discovered new evidence about the formation of distant galaxies.",
    "Space telescopes help astronomers study objects in the universe.",
    "The football team won the championship after a close final match.",
    "Athletes require regular training to improve their performance.",
    "Climate change is causing significant changes in global weather patterns.",
    "Rising temperatures are affecting ecosystems around the world.",
    "Doctors are using artificial intelligence to assist with medical diagnosis.",
    "Modern hospitals use digital systems to manage patient information.",
    "Electric vehicles are becoming more popular because they produce fewer emissions.",
    "Battery technology is important for the future of electric transportation.",
    "Cybersecurity protects computer systems from unauthorized access.",
    "Strong passwords can help users protect their online accounts.",
    "Social media platforms allow people to communicate and share information.",
    "Digital communication has changed the way people interact with each other."
]

print("Number of sentences:", len(sentences))
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

texts = sentences

embeddings = model.encode(texts)

print("Embedding shape:", embeddings.shape)
```

```python
similarity_matrix = cosine_similarity(embeddings)

print("Similarity matrix shape:", similarity_matrix.shape)
```

```python
results = []

for i in range(len(texts)):
    for j in range(i + 1, len(texts)):
        results.append({
            "Text 1": texts[i],
            "Text 2": texts[j],
            "Cosine Similarity": round(similarity_matrix[i][j], 4)
        })

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="Cosine Similarity",
    ascending=False
)

results_df.head(10)
```

```python
top_5 = results_df.head(5)

top_5
```

```python
lowest_5 = results_df.tail(5).sort_values(
    by="Cosine Similarity",
    ascending=True
)

lowest_5
```

Observations
Related sentences generally have higher cosine similarity because their meanings are closely connected.
Renewable energy and solar power sentences show high similarity.
Education and online learning sentences also have similar meanings.
Space-related sentences show similarity because they discuss the universe and astronomy.
Unrelated topics generally have lower similarity scores.

Importance of Text Embeddings
Text embeddings convert sentences into numerical vectors that represent their meaning. They help computers understand and compare the semantic relationship between different texts.
Applications in LLMs
Semantic Search: Finds relevant information based on meaning instead of exact keywords.
Question Answering: Helps retrieve relevant information to answer user questions.
Recommendation Systems: Finds similar content based on semantic meaning.
Document Similarity: Compares documents and identifies related content.
Chatbots: Helps retrieve relevant context to generate better responses.
Conclusion
This project demonstrates how text embeddings represent sentences as numerical vectors. Cosine similarity is used to measure the semantic similarity between sentence pairs. Higher scores indicate more closely related meanings, while lower scores indicate less related meanings.
